## Categorical Data :-

### 1. Countplot

```python
sns.countplot(train_df,x = 'Survived', stat='percent',hue = 'Survived',width = 0.3)
```

```python
train_df['Survived'].value_counts().plot(kind = 'bar',width = 0.3)
```

```python
sns.countplot(train_df,y = 'Pclass',hue = 'Pclass',width = 0.4)
```

```python
sns.countplot(train_df,x = 'Embarked',hue = 'Embarked',width = 0.4)
```

```python
sns.countplot(train_df,x = 'Sex',hue = 'Sex',width = 0.3)
```

### 2. Pie Chart

```python
train_df['Survived'].value_counts().plot(kind = 'pie',autopct = '%.2f',label = 'Survived counts')
```

```python
train_df['Sex'].value_counts().plot(kind = 'pie',autopct = '%.2f')
```
