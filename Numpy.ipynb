### Numpy  - Numerical Python

```python
import numpy as np
my_list = [1,2,3,4,5]
numpy_array = np.array(my_list)
print("Simple Numpy array:",numpy_array)
```

```python
zero_array = np.zeros((2,3))  # Creates an array filled with zeros.
print("Zero array:","\n",zero_array)

ones_array = np.ones((3,3))       # Creates an array filled with ones.
print("\n","ones array:","\n",ones_array)

constant_array = np.full((2,2),7)     # Creates an array filled with specific values.
print("\n","Constant array","\n",constant_array)

range_array =np.arange(0,10,2) #Start stop step
print("\n","range_array","\n",range_array)      # Creates an array filled with evenly spaced 

linspace_array = np.linspace(0,1,5)  # Start stop 
print("\n","linespace_array","\n",linspace_array)  # Creates an array filled with evenly spaced in interval 
```

```python
# random_rand - Creates array of specific shape & fills it with random values with uniform dist 

random_array = np.random.rand(2,3)
random_array
```

```python
# random_randn = Creates array of specific shape & fills it with random variables with nprma; dist.

normal_array = np.random.randn(2,2)
normal_array
```

### How to access different rows of a multidimensional NumPy array?

```python
arr = np.array([[10,20,30],
               [40,5,66],
               [70,88,94]])

print("Given Array:")
print(arr)

# Access the First & last rows of array
res_arr = arr[[0,2]]
print("\nAccessed Rows:")
print(res_arr)
```

```python
# Accessing the Middle row of 2-D NumPy array
arr = np.array([[101,20,3,10],
               [40,5,66,7],
               [70,88,9,141]
               ])
print("Given array :")
print(arr)

# Access the Middle row of array
res_arr = arr[1]
print("\n Accessed row:")
print(res_arr)
```

### Numpy stack() in Python

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

#Stacking - 
c = np.stack((a,b),axis = 0) # axis = 0 means row, axis = 1 means column
print(c)
```

```python
d = np.stack((a,b),axis = -1)
print(d)
```

```python
# stacking two 2d arrays
# input arrays
x=np.array([[1,2,3],
            [4,5,6]])

y=np.array([[7,8,9],
            [10,11,12]])
```

```python
np.stack((x,y),axis = 0)   ## ROW stack
```

```python
np.stack((x,y),axis = 1) ## Column stack
```

### Numpy Array Broadcasting

```python
'''Broadcasting in NumPy allows us to perform arithmetic
operations on arrays of different shapes without reshaping them'''

array_2d = np.array([[1,2,3],[4,5,6]])  
scaler = 10

result = array_2d +scaler
print(result)
```

```python
# 2-D array
a1 = np.array([2,4,6])
a2 = np.array([[1,3,5],[7,9,11]])
res = a1+a2
print(res)
```

```python
# Broadcasting in conditional operator

ages = np.array([12,24,35,45,60,72])
age_group = np.array(["Adult","Minor"])
result = np.where(ages>18,age_group[0],age_group[1])
print(result)
```

### NumPy - Arithmetic Operations

```python
# Addition of Arrays

a = np.array([5,72,13,100])
b = np.array([2,5,10,30])

add_ans = np.add(a,b)
print(add_ans)
```

```python
# Subtraction of Arrays

subs_ans = np.subtract(a,b)
print(subs_ans)
```

```python
# Multiplication of Arrays

mul_ans = np.multiply(a,b)
print(mul_ans)
```

```python
# Division of Arrays

div_ans = np.divide(a,b)
print(div_ans)
```

```python
pow_ans = np.power(a,b)
pow_ans
```

```python
pow_ans = np.mod(a,b)
pow_ans
```
